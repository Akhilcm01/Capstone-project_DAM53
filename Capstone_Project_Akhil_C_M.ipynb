**Feature Engineering**

Feature Engineering means creating more useful information from existing data.

```python
#Tenure

df['Tenure_Years'] = (pd.to_datetime('today') - pd.to_datetime(df['DateOfJoining'])).dt.days // 365
df.head()
```

```python
#Categorize employees

df['AgeGroup'] = pd.cut(df['Age'], bins=[18,25,35,45,60], labels=['18-25','26-35','36-45','46-60'])
```

```python
#Seniority Flag:

df['IsSenior'] = df['Age'] > 40
```

```python
df.groupby('Department')['Age'].mean()
```

```python
df.groupby('Gender')['Tenure_Years'].mean()
```

```python
#FitnessScore

df['FitnessScore'] = df['StepCount'] / df['HeartRate']
```

```python
#LoyaltyIndex

df['LoyaltyIndex'] = df['Tenure_Years'] * 10
```

```python
#WorkLifeBalanceIndex

df['WorkLifeBalanceIndex'] = df['SleepHours'] / df['WorkHours']
```

```python
df['Overtime_Status'] = np.where(
    df['WorkHours'] > 8.5,
    'Overtime (>8.5h)',
    np.where(df['WorkHours'] < 7.0, 'Under-time (<7h)', 'Standard (7-8.5h)'),
)
df['Sleep_Category'] = pd.cut(
    df['SleepHours'],
    bins=[0, 5.5, 7.5, 12],
    labels=['Sleep Deprived (<5.5h)', 'Optimal (5.5-7.5h)', 'Surplus (>7.5h)'],
)
```

```python
Overtimegroup=df.groupby("Overtime_Status")["EmployeeID"].count()
print(Overtimegroup)
```

```python
# Function
def get_avg(col):
    return df[col].mean()

# Loop
for col in ['Age', 'WorkHours', 'SleepHours']:
    print(col, ":", get_avg(col))

# Conditional
def risk_level(stress):
    if stress >= 8:
        return "High"
    elif stress >= 5:
        return "Medium"
    else:
        return "Low"

df['Risk_Level'] = df['StressLevel'].apply(risk_level)
```

**VISUALIZATION**

Data Visualization is the graphical representation of data using charts, graphs, and plots to understand patterns, trends, relationships, and outliers in a dataset.


Matplotlib

Matplotlib is a Python data visualization library that helps create various types of plots such as line charts, bar charts, histograms, scatter plots, and pie charts. It is a basic library.



```python
# 1. Scatter plot

# SleepHours vs FitnessScore

plt.scatter(df["StepCount"],df["FitnessScore"])
plt.title("StepCount vs FitnessScore")
plt.xlabel("StepCount")
plt.ylabel("FitnessScore")
plt.show()
```

Since we are finding relationship between two variables, scatter plot is the best chart we can use. The scatter plot shows strong upward trend (positive correlation). At low step counts scores are tightly clustered near the bottom (roughly 10 to 30).
At high step counts (~14000 steps)fitness score are much higher,ranging between 130 and 250+.

```python

#2 Bar plot

plt.figure(figsize=(10,6))
plt.bar(df["Department"],df["WorkLifeBalanceIndex"])
plt.title("Department vs FitnessScore")
plt.xlabel("Department")
plt.ylabel("FitnessScore")
plt.show()
```

Bar chart is best here as we are comparing categories. Unknown departments shows low work life balance and IT department shows more work life balance.

```python
#3. Hist plot


plt.hist(df['Age'], bins=10, color='skyblue', edgecolor='black')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```

The age distribution is fairly balanced across the workforce, with employees primarily ranging from 21 to 59 years. A noticeable concentration is observed in the 40-45 age group, indicating a strong presence of experienced professionals. Overall, the organization maintains a healthy mix of young, mid-career, and senior employees, supporting workforce diversity and effective succession planning.

Key Insight: The balanced age spread reflects a stable workforce with a good combination of fresh talent and experienced employees.

```python
#4. Pie chart

gender_counts = df['Gender'].value_counts()

plt.pie(gender_counts.values,
        labels=gender_counts.index,
        autopct='%1.1f%%',
        startangle=90,
        colors=['lightblue','lightpink','lightgreen','lavender'])
plt.title("Gender Distribution")
plt.show()
```

The gender distribution is highly balanced across the workforce, with Male (32.0%), Other (31.6%), and Female (31.4%) representation being nearly equal. Only 4.9% of records have gender information missing. This indicates a diverse and inclusive workforce with no significant gender imbalance.

Key Insight: The organization demonstrates strong gender diversity, which can contribute to improved collaboration, innovation, and workplace inclusivity.

```python
#5. Box plot

plt.figure()
plt.boxplot(df["HeartRate"])
plt.title("HeartRate Distribution")
plt.ylabel("HeartRate")
plt.show()
```

The box plot shows that employee heart rates are centered around a median of 90 bpm. Most heart rate values fall between approximately 75 bpm and 105 bpm, indicating moderate variability. The absence of significant outliers suggests a generally consistent heart rate distribution across the dataset.

Key Insight: Heart rates are relatively stable among employees, with no extreme values detected, indicating a broadly uniform health profile within the observed population.

```python
#6. Count plot

sns.countplot(data=df,x="SmokingStatus")
plt.show()
```

```python
#7. Voilin plot

sns.violinplot(data=df,x="Sleep_Category",y="WorkLifeBalanceIndex")
plt.show()
```

Sleep Category vs Work-Life Balance Index

This suggests that employees who get adequate sleep tend to maintain a better work-life balance.

Key Insight: Higher sleep duration is positively associated with improved work-life balance, highlighting the importance of healthy sleep habits for overall employee well-being.

```python
#8. Heatmap
plt.figure(figsize=(10,8))
corr=df.select_dtypes(include="number").corr()
sns.heatmap(corr,annot=True)
plt.show()
```

The correlation analysis indicates that physical activity (StepCount) strongly improves FitnessScore, while adequate sleep is associated with better WorkLifeBalance. Conversely, longer WorkHours significantly reduce work-life balance. These findings highlight the importance of maintaining healthy sleep habits, regular physical activity, and balanced working hours for employee well-being and productivity.
